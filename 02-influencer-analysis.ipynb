```python
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(7,7))
plt.pie(top5_fenomen["Gerceketkilesim"],labels=top5_fenomen["Ad"],autopct="%1.1f%%",startangle=140, colors=plt.cm.Pastel1.colors,wedgeprops=dict (width=0.3))
plt.title("Gerçek etkileşimi en yüksek 5 fenomen - halka grafik")
plt.show()
```
