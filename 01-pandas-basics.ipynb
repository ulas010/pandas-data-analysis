```python
import pandas as pd 
```

```python
#vize notları listesi oluşturma

vize_notlari=[50,60,20,80,100]
print("vize",vize_notlari)
```

```python
#serilerin genel kullanımı (seriler tek boyutlu bir veri yapısı , bir sütun gibi düşünebiliriz)
#series =pd.Series(data.index)
vize_serisi=pd.Series(vize_notlari)
print(vize_serisi)        
```

```python
#. max() fonksiyonu ile maksimum değeri görebiliriz 
print("en yüksek değer(max)",vize_serisi.max())
```

```python
print("en düşük değer(min)",vize_serisi.min())
```

```python
print("toplam değer(max)",vize_serisi.sum())
```

```python
print("ortalama değer(mean)",vize_serisi.mean())
```

```python
print("istatiksel özet\n",vize_serisi.describe())
```

```python
meyveler=["elma","armut","karpuz"]
fiyatlar=[30,40,50]
stok=[10,15,7]

print("meyve verileri:\n",pd.Series(meyveler))
print("fiyat verileri:\n",pd.Series(fiyatlar))
print("stok verileri:\n",pd.Series(stok))
```

```python
df=pd.DataFrame({"meyve":meyveler, "fiyat":fiyatlar, "stok":stok})
print(df)
```

```python
print(df.iloc[1,0])
```

```python
print(df.loc[2,"fiyat"])
```

```python
df=pd.DataFrame({"meyve ":meyveler, "fiyat":fiyatlar,"stok":stok}, index=[1,2,3])

print(df)
```

```python
myvindex=df.set_index("meyve")
print("indeks meyve oldu\n",myvindex)
```

```python
urunler=["kalem","defter","silgi"]
fiyat=[70,50,20]
satis_miktarı=[200,150,300]

satis=({"ürünler":urunler,"fiyatlar":fiyat, "satışlar":satis_miktarı})

print(satis)
```

```python
satis=pd.DataFrame({"ürünler":urunler,"fiyatlar":fiyat, "satışlar":satis_miktarı})
print(satis)
```

```python
print(satis.iloc[1,0])
```

```python
satis_miktarı=pd.Series(satis_miktarı)

print("en yüksek satış miktarı", satis_miktarı.max())
```

```python

satis["fiyatlar"] = [f + 10 for f in satis["fiyatlar"]]

print(satis)
```

```python
print(satis[satis["fiyatlar"]>60])
```

```python
print(satis[satis["fiyatlar"]<80])
```

```python
ortalama_fiyat =satis["fiyatlar"].mean()

print(ortalama_fiyat)
```
